```python
plt.style.use("./magma.mplstyle")
x = np.random.rand(100,100,100)
# plt.imshow(x)
plt.scatter(x[0],x[1], x[0].T*30, c= x[0]*255)
x_mod=(x[0].flatten()*100 % 20).astype(int)
colors = list(f"C{i}" for i in x_mod)
plt.grid()
plt.show()
```

```python
magma = plt.get_cmap("magma")
fig:plt.Figure = None #type:ignore
ax:plt.Axes = None #type:ignore
ax2:plt.Axes = None #type:ignore

fig, (ax,ax2) = plt.subplots(ncols=2)
fig.set_size_inches((10,10))
heatmap = ax.imshow(x[:1][0],alpha=x[2],interpolation="bicubic")
cbar = fig.colorbar(heatmap, ax=ax, shrink=0.35, location="right")
cbar.minorticks_on()

ax2.scatter(x[0], x[1], x[0]+x[1], c=x[1].T+x[0].T)
ax2.grid(True, "both", "both")


ax.set_aspect("equal")
ax2.set_aspect("equal")
plt.show()
```
